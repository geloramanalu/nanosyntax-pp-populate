```python
parser = Parser("benepar_en3")

def create_structural_cfg(sentences):
    trees = []
    for sent in sentences:
        try:
            # parse string → nltk.Tree
            t = parser.parse(sent)
            trees.append(t)
        except Exception as e:
            # optional: log or skip unparseable sentences
            print(f"  skipping (parse failed): {sent!r}  → {e!r}")

    if not trees:
        raise ValueError("No parse trees generated – check your sentences")

    # extract & dedupe productions
    all_prods   = [p for t in trees for p in t.productions()]
    unique_prods = list(dict.fromkeys(all_prods))

    # derive a start symbol that actually appears in LHS
    start_label = trees[0].label()
    start_nt    = Nonterminal(start_label)

    # build a quick lookup for reachability
    prods_by_lhs = defaultdict(list)
    for p in unique_prods:
        prods_by_lhs[p.lhs()].append(p)

    # BFS from the start symbol
    reachable = {start_nt}
    queue     = deque([start_nt])
    while queue:
        lhs = queue.popleft()
        for p in prods_by_lhs[lhs]:
            for sym in p.rhs():
                if isinstance(sym, Nonterminal) and sym not in reachable:
                    reachable.add(sym)
                    queue.append(sym)

    pruned = [p for p in unique_prods if p.lhs() in reachable]
    if not pruned:
        raise ValueError(f"No productions reachable from {start_nt!r}")

    return CFG(start=start_nt, productions=pruned)
```

```python
def load_json_cfg(json_path):
    with open(json_path, 'r') as f:
        json_cfg = json.load(f)
    productions = []
    for prod_str in json_cfg['productions']:
        lhs_str, rhs_str = prod_str.split("->")
        lhs_sym = lhs_str.strip()
        rhs_syms = rhs_str.strip().split()

        # 2. Wrap nonterminals; strip quotes from terminals
        parsed_rhs = []
        for sym in rhs_syms:
            if sym.startswith("'") and sym.endswith("'"):
                parsed_rhs.append(sym[1:-1])
            else:
                parsed_rhs.append(Nonterminal(sym))
        
        prod = Production(Nonterminal(lhs_sym), parsed_rhs)
        productions.append(prod)

    # 3. Build the CFG
    cfg = CFG(Nonterminal(json_cfg['start']), productions)
    return cfg

cfg_retrieve = load_json_cfg(json_cfg_real_no_punct_path)
```

```python
def build_pcfg_from_pos_tagged(sentences, min_prob=0.001):
    
    import nltk
    from nltk import word_tokenize, pos_tag
    
    # Download required NLTK data if not present
    try:
        nltk.data.find('tokenizers/punkt')
    except LookupError:
        nltk.download('punkt')
    
    try:
        nltk.data.find('taggers/averaged_perceptron_tagger')
    except LookupError:
        nltk.download('averaged_perceptron_tagger')
    
    production_counts = Counter()
    word_tag_counts = Counter()
    all_pos_tags = set()
    
    for sent in tqdm(sentences):
        try:
            # tokenize and POS tag
            tokens = word_tokenize(sent)
            if not tokens:
                continue
                
            pos_tags = pos_tag(tokens)
            
            
            for word, pos in pos_tags:
                all_pos_tags.add(pos)
                
                if pos.startswith('NNP'):
                    word_clean = word
                else:
                    word_clean = word.lower()
                word_tag_counts[(pos, word_clean)] += 1
            
            # rules of phrase structure
            if len(pos_tags) > 0:
                pos_sequence = [pos for word, pos in pos_tags]
                
                # S -> tags
                s_production = (Nonterminal('S'), tuple(Nonterminal(pos) for pos in pos_sequence))
                production_counts[s_production] += 1
                
       
                if len(pos_sequence) >= 2:
                    
                    np_end = -1
                    vp_start = -1
                    
                    # Find potential NP end and VP start
                    for i, pos in enumerate(pos_sequence):
                        if pos.startswith('N') and np_end == -1:
                            np_end = i
                        elif pos.startswith('V') and vp_start == -1:
                            vp_start = i
                            break
                    
                    if np_end >= 0 and vp_start > np_end:
                        # Add S -> NP VP rule
                        production_counts[(Nonterminal('S'), (Nonterminal('NP'), Nonterminal('VP')))] += 1
                        
                        # Add NP -> sequence from start to np_end
                        np_seq = pos_sequence[:np_end + 1]
                        if len(np_seq) > 0:
                            production_counts[(Nonterminal('NP'), tuple(Nonterminal(pos) for pos in np_seq))] += 1
                        
                        # Add VP -> sequence from vp_start to end
                        vp_seq = pos_sequence[vp_start:]
                        if len(vp_seq) > 0:
                            production_counts[(Nonterminal('VP'), tuple(Nonterminal(pos) for pos in vp_seq))] += 1
                
                # hardcode not reliable patterns for robustness
                # common_patterns = [
                #     # Determiner + Noun patterns
                #     (['DT', 'NN'], 'NP'),
                #     (['DT', 'NNS'], 'NP'),
                #     (['DT', 'NNP'], 'NP'),
                #     (['DT', 'JJ', 'NN'], 'NP'),
                #     (['DT', 'JJ', 'NNS'], 'NP'),
                #     # Verb patterns
                #     (['VBZ', 'DT', 'NN'], 'VP'),
                #     (['VBD', 'DT', 'NN'], 'VP'),
                #     (['VB', 'DT', 'NN'], 'VP'),
                #     # Prepositional phrases
                #     (['IN', 'DT', 'NN'], 'PP'),
                #     (['IN', 'NNP'], 'PP'),
                # ]
                
                # Check for these patterns in the sentence
                for pattern_pos, pattern_lhs in common_patterns:
                    for i in range(len(pos_sequence) - len(pattern_pos) + 1):
                        if pos_sequence[i:i+len(pattern_pos)] == pattern_pos:
                            pattern_rhs = tuple(Nonterminal(pos) for pos in pattern_pos)
                            production_counts[(Nonterminal(pattern_lhs), pattern_rhs)] += 1
                        
        except Exception as e:
            print(f"Error processing sentence: {e}")
            continue
    
    
    prob_productions = []
    
    # Word -> POS productions
    pos_counts = defaultdict(int)
    for (pos, word), count in word_tag_counts.items():
        pos_counts[pos] += count
    
    # Add all word productions
    for (pos, word), count in word_tag_counts.items():
        prob = count / pos_counts[pos]
        prob_prod = ProbabilisticProduction(Nonterminal(pos), [word], prob=prob)
        prob_productions.append(prob_prod)
    
    
    common_pos_tags = ['NN', 'NNS', 'NNP', 'NNPS', 'VB', 'VBD', 'VBG', 'VBN', 'VBP', 'VBZ', 
                       'JJ', 'JJR', 'JJS', 'RB', 'RBR', 'RBS', 'DT', 'IN', 'CC', 'TO', 'PRP', 'MD']
    
    unknown_prob = 0.0001  #  low probability for unknown words
    for pos in common_pos_tags:
        if pos in all_pos_tags:  
            
            prob_prod = ProbabilisticProduction(Nonterminal(pos), ['<UNK>'], prob=unknown_prob)
            prob_productions.append(prob_prod)
    
    # phrase structure productions
    lhs_counts = defaultdict(int)
    for (lhs, rhs), count in production_counts.items():
        lhs_counts[lhs] += count
    
    for (lhs, rhs), count in production_counts.items():
        prob = count / lhs_counts[lhs]
        if prob >= min_prob:
            prob_prod = ProbabilisticProduction(lhs, list(rhs), prob=prob)
            prob_productions.append(prob_prod)
    
    if len([p for p in prob_productions if not all(isinstance(sym, str) for sym in p.rhs())]) < 10:
        print("Adding fallback structural rules...")
        # temp
        # fallback_rules = [
        #     (Nonterminal('S'), [Nonterminal('NP')], 0.1),
        #     (Nonterminal('S'), [Nonterminal('VP')], 0.1),
        #     (Nonterminal('NP'), [Nonterminal('NN')], 0.3),
        #     (Nonterminal('NP'), [Nonterminal('NNP')], 0.3),
        #     (Nonterminal('VP'), [Nonterminal('VB')], 0.3),
        #     (Nonterminal('VP'), [Nonterminal('VBZ')], 0.3),
        # ]
        
        for lhs, rhs, prob in fallback_rules:
            prob_prod = ProbabilisticProduction(lhs, rhs, prob=prob)
            prob_productions.append(prob_prod)
    
    print(f"Robust PCFG: {len(prob_productions)} productions")
    print(f"Lexical productions: {len([p for p in prob_productions if all(isinstance(sym, str) for sym in p.rhs())])}")
    print(f"Structural productions: {len([p for p in prob_productions if not all(isinstance(sym, str) for sym in p.rhs())])}")
    
    return PCFG(Nonterminal('S'), prob_productions)


def parse_with_unknown_word_handling(parser, tokens, pcfg):
    
    grammar_words = set()
    for prod in pcfg.productions():
        if all(isinstance(sym, str) for sym in prod.rhs()):
            grammar_words.update(prod.rhs())
    
    # replace unknown words with <UNK>
    processed_tokens = []
    for token in tokens:
        if token.lower() in grammar_words or token in grammar_words:
            processed_tokens.append(token.lower() if not token[0].isupper() else token)
        else:
            processed_tokens.append('<UNK>')
    
    return list(parser.parse(processed_tokens))
```

```python
from nltk import CFG, Nonterminal, Production

def compress_cfg(cfg: CFG) -> CFG:
    """
    Eliminate chain (unit) productions from an NLTK CFG by
    replacing A -> B with productions A -> α for each B -> α
    non-chain production, effectively 'neglecting' the intermediate B.
    
    Args:
        cfg (nltk.grammar.CFG): Input context-free grammar.
    
    Returns:
        nltk.grammar.CFG: A new CFG with chain productions removed.
    """
    start = cfg.start()
    productions = cfg.productions()

    # Identify chain (unit) vs. non-chain productions
    chain_prods     = [p for p in productions if len(p.rhs()) == 1 and isinstance(p.rhs()[0], Nonterminal)]
    non_chain_prods = [p for p in productions if p not in chain_prods]

    # Collect all nonterminals
    nonterminals = set(p.lhs() for p in productions)
    for p in productions:
        for sym in p.rhs():
            if isinstance(sym, Nonterminal):
                nonterminals.add(sym)

    # Build chain relations: A -> B
    chain_map = {}
    for p in chain_prods:
        A, B = p.lhs(), p.rhs()[0]
        chain_map.setdefault(A, set()).add(B)

    # Compute, for each A, the set of nonterminals reachable via chain productions
    closure = {}
    for A in nonterminals:
        reached = set()
        stack = [A]
        while stack:
            X = stack.pop()
            for Y in chain_map.get(X, []):
                if Y not in reached:
                    reached.add(Y)
                    stack.append(Y)
        closure[A] = reached

    # Build the new production set:
    new_prods = set(non_chain_prods)
    for A, Bs in closure.items():
        for B in Bs:
            # For each non-chain B -> α, add A -> α
            for p in non_chain_prods:
                if p.lhs() == B:
                    new_prods.add(Production(A, p.rhs()))

    # Return the new CFG
    return CFG(start, list(new_prods))




compressed_cfg_real_no_punct___ = compress_cfg(cfg_real_no_punct)

print("Original productions:")
for p in cfg_real_no_punct.productions():
    print("  ", p)
print("\nCompressed productions:")
for p in compressed_cfg_real_no_punct___.productions():
    print("  ", p)
    
print(len(cfg_real_no_punct.productions())) #2785
print(len(compressed_cfg_real_no_punct___.productions())) #18332
```

```python
import nltk
from nltk.grammar import CFG, Production, Nonterminal
from collections import defaultdict, Counter
import time
from typing import Set, Dict, List, Tuple

class GrammarOptimizer:
    def __init__(self, cfg: CFG):
        self.cfg = cfg
        self.productions = list(cfg.productions())
        self.start_symbol = cfg.start()
        
    def optimize_for_earley(self, 
                           remove_epsilon=True,
                           remove_unreachable=True, 
                           remove_unproductive=True,
                           factorize_left=True,
                           merge_equivalent=True,
                           optimize_lexical=True) -> CFG:
        """
        Comprehensive grammar optimization for Earley parsing speed
        """
        print(f"Starting optimization with {len(self.productions)} productions")
        
        optimized_productions = self.productions.copy()
        
        # Step 1: Remove epsilon productions (if any)
        if remove_epsilon:
            optimized_productions = self._remove_epsilon_productions(optimized_productions)
            print(f"After epsilon removal: {len(optimized_productions)} productions")
        
        # Step 2: Remove unreachable symbols
        if remove_unreachable:
            optimized_productions = self._remove_unreachable_symbols(optimized_productions)
            print(f"After unreachable removal: {len(optimized_productions)} productions")
        
        # Step 3: Remove unproductive symbols
        if remove_unproductive:
            optimized_productions = self._remove_unproductive_symbols(optimized_productions)
            print(f"After unproductive removal: {len(optimized_productions)} productions")
        
        # Step 4: Left factorization (reduces backtracking)
        if factorize_left:
            optimized_productions = self._left_factorize(optimized_productions)
            print(f"After left factorization: {len(optimized_productions)} productions")
        
        # Step 5: Merge equivalent productions
        if merge_equivalent:
            optimized_productions = self._merge_equivalent_productions(optimized_productions)
            print(f"After merging equivalent: {len(optimized_productions)} productions")
        
        # Step 6: Optimize lexical productions
        if optimize_lexical:
            optimized_productions = self._optimize_lexical_productions(optimized_productions)
            print(f"After lexical optimization: {len(optimized_productions)} productions")
        
        return CFG(self.start_symbol, optimized_productions)
    
    def _remove_epsilon_productions(self, productions: List[Production]) -> List[Production]:
        """Remove epsilon productions which slow down Earley parsing"""
        # For now, just filter out any empty productions
        return [p for p in productions if len(p.rhs()) > 0]
    
    def _remove_unreachable_symbols(self, productions: List[Production]) -> List[Production]:
        """Remove symbols that cannot be reached from start symbol"""
        reachable = {self.start_symbol}
        changed = True
        
        while changed:
            changed = False
            for prod in productions:
                if prod.lhs() in reachable:
                    for symbol in prod.rhs():
                        if isinstance(symbol, Nonterminal) and symbol not in reachable:
                            reachable.add(symbol)
                            changed = True
        
        return [p for p in productions if p.lhs() in reachable]
    
    def _remove_unproductive_symbols(self, productions: List[Production]) -> List[Production]:
        """Remove symbols that cannot derive terminal strings"""
        productive = set()
        
        # First, find all productions that derive only terminals
        for prod in productions:
            if all(isinstance(symbol, str) for symbol in prod.rhs()):
                productive.add(prod.lhs())
        
        # Iteratively find productive nonterminals
        changed = True
        while changed:
            changed = False
            for prod in productions:
                if (prod.lhs() not in productive and 
                    all(isinstance(symbol, str) or symbol in productive 
                        for symbol in prod.rhs())):
                    productive.add(prod.lhs())
                    changed = True
        
        return [p for p in productions 
                if p.lhs() in productive and 
                all(isinstance(symbol, str) or symbol in productive 
                    for symbol in prod.rhs())]
    
    def _left_factorize(self, productions: List[Production]) -> List[Production]:
        """
        Left factorization reduces backtracking in Earley parsing
        A -> αβ₁ | αβ₂ becomes A -> αA', A' -> β₁ | β₂
        """
        # Group productions by LHS
        lhs_groups = defaultdict(list)
        for prod in productions:
            lhs_groups[prod.lhs()].append(prod)
        
        new_productions = []
        counter = 0
        
        for lhs, prods in lhs_groups.items():
            if len(prods) <= 1:
                new_productions.extend(prods)
                continue
            
            # Group by common prefixes
            prefix_groups = defaultdict(list)
            for prod in prods:
                if len(prod.rhs()) > 0:
                    prefix_groups[prod.rhs()[0]].append(prod)
                else:
                    new_productions.append(prod)
            
            for prefix, prefix_prods in prefix_groups.items():
                if len(prefix_prods) <= 1:
                    new_productions.extend(prefix_prods)
                    continue
                
                # Check for longer common prefixes
                common_prefix = [prefix]
                min_len = min(len(p.rhs()) for p in prefix_prods)
                
                for i in range(1, min_len):
                    if all(p.rhs()[i] == prefix_prods[0].rhs()[i] for p in prefix_prods):
                        common_prefix.append(prefix_prods[0].rhs()[i])
                    else:
                        break
                
                if len(common_prefix) > 1 or len(prefix_prods) > 2:
                    # Create new nonterminal
                    new_nt = Nonterminal(f"{lhs.symbol()}_LF{counter}")
                    counter += 1
                    
                    # A -> αA'
                    new_productions.append(Production(lhs, common_prefix + [new_nt]))
                    
                    # A' -> β₁ | β₂ | ...
                    for prod in prefix_prods:
                        suffix = prod.rhs()[len(common_prefix):]
                        if suffix:
                            new_productions.append(Production(new_nt, suffix))
                        else:
                            # Handle case where some production is just the prefix
                            new_productions.append(Production(new_nt, []))
                else:
                    new_productions.extend(prefix_prods)
        
        return new_productions
    
    def _merge_equivalent_productions(self, productions: List[Production]) -> List[Production]:
        """Merge productions with identical RHS"""
        # Group by RHS
        rhs_groups = defaultdict(list)
        for prod in productions:
            rhs_key = tuple(prod.rhs())
            rhs_groups[rhs_key].append(prod)
        
        merged_productions = []
        for rhs_key, prods in rhs_groups.items():
            if len(prods) > 1:
                # Keep only one production per unique RHS
                # Choose the LHS that appears most frequently in the grammar
                lhs_counts = Counter(p.lhs() for p in prods)
                best_lhs = lhs_counts.most_common(1)[0][0]
                merged_productions.append(Production(best_lhs, list(rhs_key)))
            else:
                merged_productions.extend(prods)
        
        return merged_productions
    
    def _optimize_lexical_productions(self, productions: List[Production]) -> List[Production]:
        """
        Optimize lexical productions for faster lookup
        Group terminal productions by POS tags where possible
        """
        lexical_prods = []
        non_lexical_prods = []
        
        # Separate lexical from non-lexical productions
        for prod in productions:
            if (len(prod.rhs()) == 1 and isinstance(prod.rhs()[0], str)):
                lexical_prods.append(prod)
            else:
                non_lexical_prods.append(prod)
        
        # Group lexical productions by terminal
        terminal_groups = defaultdict(list)
        for prod in lexical_prods:
            terminal_groups[prod.rhs()[0]].append(prod)
        
        optimized_lexical = []
        for terminal, prods in terminal_groups.items():
            if len(prods) > 1:
                # If multiple POS tags for same word, keep most specific ones
                # Priority: specific POS > general POS
                pos_priority = {
                    'NNP': 10,   # Proper noun (most specific)
                    'NN': 8,     # Common noun
                    'VBN': 9,    # Past participle
                    'VBD': 8,    # Past tense
                    'VB': 6,     # Base verb
                    'JJ': 7,     # Adjective
                    'DT': 9,     # Determiner
                    'IN': 8,     # Preposition
                }
                
                # Sort by priority and keep top ones
                sorted_prods = sorted(prods, 
                                    key=lambda p: pos_priority.get(p.lhs().symbol(), 5), 
                                    reverse=True)
                
                # Keep at most 2-3 POS tags per word to reduce ambiguity
                optimized_lexical.extend(sorted_prods[:2])
            else:
                optimized_lexical.extend(prods)
        
        return non_lexical_prods + optimized_lexical
    
    def create_lookup_optimized_grammar(self, productions: List[Production]) -> CFG:
        """
        Create a grammar optimized for fast symbol lookup during parsing
        """
        # Sort productions to put more specific/likely ones first
        def production_priority(prod):
            score = 0
            
            # Lexical productions get higher priority
            if len(prod.rhs()) == 1 and isinstance(prod.rhs()[0], str):
                score += 100
            
            # Shorter productions generally parse faster
            score += 50 / (len(prod.rhs()) + 1)
            
            # Common non-terminals get higher priority
            common_nts = {'NP', 'VP', 'PP', 'ADJP', 'S', 'SBAR'}
            if prod.lhs().symbol() in common_nts:
                score += 20
            
            return score
        
        sorted_productions = sorted(productions, key=production_priority, reverse=True)
        return CFG(self.start_symbol, sorted_productions)

def benchmark_parser(parser, test_sentences, max_time=30):
    """Benchmark parser performance"""
    results = []
    
    for i, sentence in enumerate(test_sentences):
        if i >= 2:  # Limit testing to avoid long waits
            break
            
        tokens = nltk.word_tokenize(sentence)
        start_time = time.time()
        
        try:
            # Try to get first parse with timeout
            parses = list(parser.parse(tokens))
            parse_time = time.time() - start_time
            
            if parse_time > max_time:
                results.append((sentence, None, parse_time, "TIMEOUT"))
            else:
                results.append((sentence, len(parses), parse_time, "SUCCESS"))
                
        except Exception as e:
            parse_time = time.time() - start_time
            results.append((sentence, None, parse_time, f"ERROR: {str(e)}"))
    
    return results

# Main optimization function
def optimize_grammar_for_speed(cfg: CFG, test_sentences=None) -> CFG:
    """
    Main function to optimize grammar for Earley parsing speed
    """
    optimizer = GrammarOptimizer(cfg)
    
    print("=== Grammar Optimization for Earley Parser ===")
    print(f"Original grammar: {len(cfg.productions())} productions")
    
    # Basic optimization
    optimized_cfg = optimizer.optimize_for_earley(
        remove_epsilon=True,
        remove_unreachable=True,
        remove_unproductive=True,
        factorize_left=True,
        merge_equivalent=True,
        optimize_lexical=True
    )
    
    print(f"Optimized grammar: {len(optimized_cfg.productions())} productions")
    
    # Create lookup-optimized version
    final_cfg = optimizer.create_lookup_optimized_grammar(optimized_cfg.productions())
    
    # Benchmark if test sentences provided
    if test_sentences:
        print("\n=== Performance Comparison ===")
        
        original_parser = nltk.parse.EarleyChartParser(cfg)
        optimized_parser = nltk.parse.EarleyChartParser(final_cfg)
        
        print("Testing original grammar...")
        print('skip origin parser')
        # original_results = benchmark_parser(original_parser, test_sentences)
        
        print("Testing optimized grammar...")
        optimized_results = benchmark_parser(optimized_parser, test_sentences)
        
        # Compare results
        # print(f"\n{'Sentence':<40} {'Original':<10} {'Optimized':<10} {'Speedup':<10}")
        # print("-" * 80)
        
        # for i, (orig, opt) in enumerate(zip(original_results, optimized_results)):
        #     orig_time = orig[2] if orig[3] == "SUCCESS" else float('inf')
        #     opt_time = opt[2] if opt[3] == "SUCCESS" else float('inf')
        #     speedup = orig_time / opt_time if opt_time > 0 and opt_time != float('inf') else 0
            
        #     sentence = orig[0][:37] + "..." if len(orig[0]) > 40 else orig[0]
        #     print(f"{sentence:<40} {orig_time:<10.2f} {opt_time:<10.2f} {speedup:<10.2f}x")
        print("\n=== Benchmark Results ===")
        for sentence, count, time_taken, status in optimized_results:
            if count is not None:
                print(f"Sentence: {sentence!r} → {count} parses in {time_taken:.2f}s [{status}]")
            else:
                print(f"Sentence: {sentence!r} → {status} in {time_taken:.2f}s")
        
    return final_cfg

# Usage example:

# Load your original CFG
# cfg = nltk.data.load('your_grammar.cfg')

# Optimize the grammar
optimized_cfg = optimize_grammar_for_speed(cfg_real_no_punct, unique_split_annotations_filtered[:3])

# Create optimized parser
# optimized_parser = nltk.parse.EarleyChartParser(optimized_cfg)
```

```python
# Define the language for the sentence as well as for the spaCy and benepar models
language = Language.English
nlp = ConstituentTree.create_pipeline(language, ConstituentTree.SpacyModelSize.Small)

def clean_text_aggressively(text):
    """More aggressive text cleaning to help language detection"""
    # Replace smart quotes and special characters
    cleaned = text.replace(''', "'").replace(''', "'").replace('"', '"').replace('"', '"')
    
    # Replace problematic characters that might confuse language detection
    cleaned = cleaned.replace('+', 'and')  # Replace + with 'and'
    cleaned = re.sub(r'\d+', 'NUM', cleaned)  # Replace numbers with 'NUM'
    
    # Ensure proper sentence structure
    if not cleaned.strip().endswith(('.', '!', '?')):
        cleaned = cleaned.strip() + '.'
    
    # Remove extra spaces
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    
    return cleaned

def make_text_more_english(text):
    """Make text more obviously English to help language detection"""
    # Add common English words to help detection
    if not any(word in text.lower() for word in ['the', 'a', 'an', 'is', 'was', 'are', 'were']):
        # Prepend with "This is" to make it more clearly English
        return f"This is {text.lower()}"
    return text

def make_pos_cfg_and_parse_v1(text):
    """Version 1: Aggressive text cleaning"""
    print(f"Original text: {text}")
    
    # Clean the text aggressively
    cleaned_text = clean_text_aggressively(text)
    print(f"Cleaned text: {cleaned_text}")
    
    # Tokenize & POS-tag
    doc = nlp(cleaned_text)
    tokens = [tok.text for tok in doc]
    pos_seq = [tok.tag_ for tok in doc]
    
    print(f"Tokens: {tokens}")
    print(f"POS tags: {pos_seq}")
    
    # Build CTL tree without the sentence_language parameter
    ctl = ConstituentTree(doc, nlp, Structure.WithoutTokenLeaves)
    
    # Extract productions and make CFG
    tree = ctl.tree
    prods = tree.productions()
    G = CFG(Nonterminal("S"), prods)
    
    print(f"Grammar productions: {len(prods)}")
    
    # Parse the POS sequence
    parser = EarleyChartParser(G)
    try:
        G.check_coverage(pos_seq)
    except ValueError as e:
        print(f"Grammar coverage issue: {e}")
        terminals = set()
        for prod in prods:
            if prod.is_lexical():
                terminals.add(prod.rhs()[0])
        print(f"Available POS tags in grammar: {sorted(terminals)}")
        print(f"Required POS tags: {pos_seq}")
        raise RuntimeError(f"Missing POS in lexicon: {e}")
    
    parses = list(parser.parse(pos_seq))
    if not parses:
        raise RuntimeError("No parse found!")
    
    print(f"Found {len(parses)} parse(s)")
    parse = parses[0]
    
    # Re-attach original tokens (not cleaned ones)
    original_doc = nlp(text.replace(''', "'").replace(''', "'").replace('"', '"').replace('"', '"'))
    original_tokens = [tok.text for tok in original_doc]
    
    i = 0
    for subtree in parse.subtrees():
        if subtree.height() == 2 and i < len(original_tokens):
            subtree[0] = original_tokens[i]
            i += 1
    
    return parse

def make_pos_cfg_and_parse_v2(text):
    """Version 2: Make text more obviously English"""
    print(f"Original text: {text}")
    
    # Make text more English-like
    english_text = make_text_more_english(text)
    cleaned_text = clean_text_aggressively(english_text)
    print(f"English-enhanced text: {cleaned_text}")
    
    # Tokenize & POS-tag
    doc = nlp(cleaned_text)
    tokens = [tok.text for tok in doc]
    pos_seq = [tok.tag_ for tok in doc]
    
    print(f"Tokens: {tokens}")
    print(f"POS tags: {pos_seq}")
    
    # Build CTL tree
    ctl = ConstituentTree(doc, nlp, Structure.WithoutTokenLeaves)
    
    # Extract productions and make CFG
    tree = ctl.tree
    prods = tree.productions()
    G = CFG(Nonterminal("S"), prods)
    
    print(f"Grammar productions: {len(prods)}")
    
    # Parse
    parser = EarleyChartParser(G)
    try:
        G.check_coverage(pos_seq)
    except ValueError as e:
        print(f"Grammar coverage issue: {e}")
        terminals = set()
        for prod in prods:
            if prod.is_lexical():
                terminals.add(prod.rhs()[0])
        print(f"Available POS tags in grammar: {sorted(terminals)}")
        print(f"Required POS tags: {pos_seq}")
        raise RuntimeError(f"Missing POS in lexicon: {e}")
    
    parses = list(parser.parse(pos_seq))
    if not parses:
        raise RuntimeError("No parse found!")
    
    parse = parses[0]
    
    # Re-attach tokens
    i = 0
    for subtree in parse.subtrees():
        if subtree.height() == 2:
            subtree[0] = tokens[i]
            i += 1
    
    return parse

def make_pos_cfg_and_parse_v3(text):
    """Version 3: Use simple, known English sentences for grammar extraction"""
    print(f"Target text: {text}")
    
    # Use a simple English sentence to extract grammar
    simple_sentence = "The building is traditional."
    print(f"Using simple sentence for grammar: {simple_sentence}")
    
    # Get grammar from simple sentence
    simple_doc = nlp(simple_sentence)
    ctl_simple = ConstituentTree(simple_doc, nlp, Structure.WithoutTokenLeaves)
    tree_simple = ctl_simple.tree
    prods = tree_simple.productions()
    
    # Now parse the target text
    cleaned_text = clean_text_aggressively(text)
    doc = nlp(cleaned_text)
    tokens = [tok.text for tok in doc]
    pos_seq = [tok.tag_ for tok in doc]
    
    print(f"Target tokens: {tokens}")
    print(f"Target POS tags: {pos_seq}")
    
    G = CFG(Nonterminal("S"), prods)
    print(f"Grammar productions: {len(prods)}")
    
    parser = EarleyChartParser(G)
    try:
        G.check_coverage(pos_seq)
    except ValueError as e:
        print(f"Grammar coverage issue: {e}")
        # Try to add missing productions
        print("Attempting to extend grammar...")
        raise RuntimeError(f"Missing POS in lexicon: {e}")
    
    parses = list(parser.parse(pos_seq))
    if not parses:
        raise RuntimeError("No parse found!")
    
    parse = parses[0]
    
    i = 0
    for subtree in parse.subtrees():
        if subtree.height() == 2:
            subtree[0] = tokens[i]
            i += 1
    
    return parse

# Test different approaches
test_sentences = [
    "Pub called Deacon Brodie's Tavern.",
    "Traditional building with windows.",
    "The cat sat on the mat.",
    "A red door.",
]

approaches = [
    ("Aggressive Cleaning", make_pos_cfg_and_parse_v1),
    ("English Enhancement", make_pos_cfg_and_parse_v2),
    ("Simple Grammar", make_pos_cfg_and_parse_v3),
]

for sentence in test_sentences:
    print(f"\n{'='*60}")
    print(f"TESTING SENTENCE: {sentence}")
    print('='*60)
    
    success = False
    for approach_name, approach_func in approaches:
        print(f"\n--- Trying {approach_name} ---")
        try:
            parsed_tree = approach_func(sentence)
            print("SUCCESS!")
            print(parsed_tree.pformat(margin=60))
            success = True
            break
        except Exception as e:
            print(f"Failed: {e}")
            continue
    
    if success:
        print(f"\n✓ Successfully parsed: {sentence}")
        break
    else:
        print(f"\n✗ All approaches failed for: {sentence}")

print(f"\n{'='*60}")
print("RECOMMENDATION:")
print("If none of these work, consider:")
print("1. Updating constituent_treelib: pip install --upgrade constituent-treelib")
print("2. Using a different parsing library like spacy-stanza")
print("3. Using NLTK's built-in parsers directly")
print('='*60)
```
